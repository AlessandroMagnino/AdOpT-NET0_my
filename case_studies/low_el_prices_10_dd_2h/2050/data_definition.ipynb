```python
import pandas as pd
import numpy as np
from pathlib import Path
import os

# CO2 tax inputs
carbon_costs_dict = {
    'price': 200, # EUR/tCO2
    'subsidy': 0 # EUR/tCO2
}

# Electricity grid inputs
electricity_grid_dict = {
    'demand': 0.0, # MW
    'scenario': 'flexible', # flexible, flexible-moderate, rigid, rigid-tes
    'resolution': '2h', # 1h or 2h
    'year': '2050', # 2025, 2030, 2040, 2050
    'export_price': 0.01, # EUR/MWh
}

# Electrcity inputs
electricity_dict = {
    'demand': 0.0, # MW
    'import_price': np.nan, # EUR/MWh
    'export_price': 0.01, # EUR/MWh
    'import_limit': 0, # MW, everythin passes through the grid connection technology
    'export_limit': 1e3, # MW
}

# Hydrogen inputs
hydrogen_dict = {
    'demand': 0.0, # t/h
    'import_price': np.nan, # EUR/t
    'export_price': 0.01, # EUR/t
    'import_limit': 0, # t/h
    'export_limit': 300 # t/h
}

# Hydrogen transported inputs
hydrogen_transported_dict = {
    'demand': 0.0, # t/h
    'import_price': np.nan, # EUR/t
    'export_price': 0.01, # EUR/t
    'import_limit': 0, # t/h
    'export_limit': 300 # t/h
}

# Ammonia inputs
ammonia_dict = {
    'import_price': np.nan, # EUR/t
    'import_limit': 0, # t/h
    'export_price': 0.01, # EUR/t
    'export_limit': 100 # t/h
}

# Ethylene inputs
ethylene_dict = {
    'import_price': np.nan, # EUR/t
    'import_limit': 0, # t/h
    'export_price': 0.01, # EUR/t
    'export_limit': 200, # t/h
}

# Propylene inputs
propylene_dict = {
    'import_price': np.nan, # EUR/t
    'import_limit': 0, # t/h
    'export_price': 0.01, # EUR/t
    'export_limit': 200, # t/h
}

# Methanol inputs
methanol_dict = {
    'import_price': np.nan, # EUR/t
    'import_limit': 0, # t/h
    'export_price': 0.01, # EUR/t
    'export_limit': 100, # t/h
}

# CO2 inputs
CO2_dict = {
    'demand': 0, # t/h
    'import_price': np.nan, # EUR/t
    'import_limit': 0, # t/h
    'export_price': -98, # EUR/t
    'export_limit': {
        'NL1': 11, # Mt/y, convert to t/h in the code
        'NL2': 0,
        'NL3': 0,
        'NL4': 11, # Mt/y, convert to t/h in the code
        'DE': 0,
        'BEL1': 2.5, # Mt/y, convert to t/h in the code
        'BEL2': 0
    },
    'export_emission_factor': -1 # tCO2/t feedstock
}

# Methane inputs
methane_dict = {
    'demand': 0, # t/h
    'year': '2050', # 2025, 2030, 2040, 2050
    'export_price': 0, # EUR/t
    'import_limit': 1000, # t/h
    'export_limit': 100, # t/h
    'import_emission_factor': 2.78 # tCO2/t feedstock, from Julia's
}

# # Heat inputs
# heat_dict = {
#     'import_price': np.nan, # EUR/MWh
#     'demand': 0.0, # MW
#     'export_price': 0.01, # EUR/MWh
#     'import_limit': 0, # MW
#     'export_limit': 1e6, # MW
# }

# Steam inputs
steam_dict = {
    'import_price': np.nan, # EUR/MWh
    'demand': 0.0, # MW
    'export_price': 0, # EUR/MWh
    'import_limit': 0, # MW
    'export_limit': 1000, # MW
}

# biomass feedstock inputs
biomass_dict = {
    'scenario': 'ENS_Med', # ENS_Low, ENS_Med, ENS_High
    'year': '2050', # 2020, 2030, 2040, 2050
    'demand': 0.0, # t/h
    'export_price': 0.01, # EUR/t
    'export_limit': 1000, # t/h
    'import_limit_share': 50, # in % of the total availability according to the scenario
    'import_emission_factor': 0,
}

methane_bio_dict = {
    'year': '2050', # 2025, 2030, 2040, 2050
    'import_price': 1833, # EUR/t (from 141 €/MWh from Julia)
    'demand': 0.0, # t/h
    'export_price': 0.01, # EUR/t
    'import_limit_share': 100, # in % of the total methane bio potential
    'export_limit': 100, # t/h
}

feedgas_dict = {
    'import_price': np.nan, # EUR/t
    'export_price_percentage': 50, # in % of the import price of methane, only for export
    'demand': 0.0, # t/h
    'import_limit': 0, # t/h
    'export_limit': 500, # t/h
}

syngas_dict = {
    'import_price': np.nan, # EUR/t
    'demand': 0.0, # t/h
    'export_price': 0.01, # EUR/t
    'import_limit': 0, # t/h
    'export_limit': 100, # t/h
}

crackergas_dict = {
    'import_price': np.nan, # EUR/t
    'demand': 0.0, # t/h
    'export_price': 0.01, # EUR/t
    'import_limit': 0, # t/h
    'export_limit': 100, # t/h
}

olefins_dict = {
    'import_price': np.nan, # EUR/t
    'demand': 0.0, # t/h
    'export_price': 0.01, # EUR/t
    'import_limit': 0, # t/h
    'export_limit': 100, # t/h
}

naphtha_dict = {
    'import_price': 732, # EUR/t
    'demand': 0.0, # t/h
    'export_price': 0.01, # EUR/t
    'import_limit': 5000, # t/h
    'export_limit': 100, # t/h
    'import_emission_factor': 3.23 # tCO2/t feedstock, from Julia's
}

HBfeed_dict = {
    'import_price': np.nan, # EUR/t
    'demand': 0.0, # t/h
    'export_price': 0.01, # EUR/t
    'import_limit': 0, # t/h
    'export_limit': 1000, # t/h
}

nitrogen_dict = {
    'import_price': np.nan, # EUR/t
    'demand': 0.0, # t/h
    'export_price': 0.01, # EUR/t
    'import_limit': 0, # t/h
    'export_limit': 1000, # t/h
}

MPW_dict = {
    'scenario': 'PP', # PP or DA
    'year': '2050', # 2022, 2030, 2040, 2050
    'demand': 0.0, # t/h
    'export_price': 0.01, # EUR/t
    'export_limit': 1000, # t/h
    'import_limit_share': 50 # in % of the total MPW production
}

naphtha_bio_dict = {
    'year': '2050', # 2025, 2030, 2040, 2050
    'import_price': 1830, # EUR/t from Julia's
    'demand': 0.0, # t/h
    'export_price': 0.01, # EUR/t
    'export_limit': 100, # t/h
    'import_limit_scenario': 'availability_low', # availability_low, availability_high
}

CO2captured_dict = {
    'demand': 0.0, # t/h
    'import_price': 0, # EUR/t
    'export_price': 0, # EUR/t
    'import_limit': 0, # t/h
    'export_limit': 0, # t/h
}

# CO2_DAC_dict = {
#     'demand': 0.0, # t/h
#     'import_price': 0,
#     'export_price': 0.01, # EUR/t
#     'import_limit': 0, # t/h
#     'export_limit': 0, # t/h
# }

cracking_byproducts_dict = {
    'import_price': np.nan, # EUR/t
    'demand': 0.0, # t/h
    'export_price': 591, # EUR/t, 0.191/0.391*0.5*p_naphtha + 0.2/0.391*1.1*p_naphtha
    'import_limit': 0, # t/h
    'export_limit': 1000, # t/h
}

def carbon_costs_definition(carbon_costs_dict, resolution):

    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    with pd.ExcelWriter(Path(carriers_path) / 'carbon_costs.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'price': carbon_costs_dict['price'], # EUR/tCO2
                'subsidy': carbon_costs_dict['subsidy'] # EUR/tCO2
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def electricity_grid_carrier(electricity_dict, resolution):

    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    country_dict = {
    'Belgium': 'BE',
    'Germany': 'DE',
    'Netherlands': 'NL'}

    position_dict = {
        'BEL1': 'coastal',
        'BEL2': 'inland',
        'DE': 'inland',
        'NL1': 'coastal',
        'NL2': 'inland',
        'NL3': 'coastal',
        'NL4': 'coastal'}

    # Read data
    if electricity_dict['resolution'] == '1h':
        if electricity_dict['year'] == '2025':
            prices_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/prices_AC_allin_2025.csv', sep=',', index_col=0)
            carbon_intensity_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/carbon_intensity_AC_net_2025.csv', sep=',', index_col=0)
        else:
            prices_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/prices_AC_allin_{electricity_dict["scenario"]}_{electricity_dict["year"]}.csv', sep=',', index_col=0)
            carbon_intensity_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/carbon_intensity_AC_net_{electricity_dict["scenario"]}_{electricity_dict["year"]}.csv', sep=',', index_col=0)
    elif electricity_dict['resolution'] == '2h':
        if electricity_dict['year'] == '2025':
            prices_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/prices_AC_allin_2025_2h.csv', sep=',', index_col=0)
            carbon_intensity_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/carbon_intensity_AC_net_2025_2h.csv', sep=',', index_col=0)
        else:
            prices_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/prices_AC_allin_{electricity_dict["scenario"]}_{electricity_dict["year"]}_2h.csv', sep=',', index_col=0)
            carbon_intensity_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/carbon_intensity_AC_net_{electricity_dict["scenario"]}_{electricity_dict["year"]}_2h.csv', sep=',', index_col=0)
    import_limit_data = pd.read_excel(project_folder / f'data/carriers_data/electricity/import_limits.xlsx', sheet_name='power_limits', index_col=0)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'electricity_grid.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:
            country = nodes.loc[plant, 'country']
            country_code = country_dict[country]

            # Prices
            prices = prices_data[country_code].values
            prices = np.clip(prices, 0.01, None)  # avoid negative prices

            # Carbon intensity
            ci = carbon_intensity_data[country_code].values / 1000 # convert from kgCO2/MWh to tCO2/MWh
            ci = np.clip(ci, 0, None)  # avoid negative carbon intensity
            ci[ci < 1e-3] = 0 # set very low carbon intensity to 0 to avoid numerical issues in the optimization

            # Import limits
            # import_limit_data is a df with years as index and inland/coastal as columns, we take the row corresponding to the year of interest and the column corresponding to the position of the plant
            position = position_dict[plant]
            import_limit = import_limit_data.loc[int(electricity_dict['year']), position]

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': prices,
                'demand': electricity_dict['demand'] * timestep,     # MW
                'export_price': electricity_dict['export_price'],    # EUR/MWh
                'import_limit': import_limit * timestep ,    # MW
                'export_limit': import_limit * timestep,    # MW
                'import_emission_factor': ci # tCO2/MWh
            }, index=full_index)

            # Save to excel file in case study folder in dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def electricity_carrier(electricity_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'electricity.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'electricity']/8760 if 'electricity' in nodes.columns else electricity_dict['demand'] # convert from GWh/y to MW

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': electricity_dict['import_price'], # EUR/MWh
                'demand': demand * timestep,     # MW
                'export_price': electricity_dict['export_price'], # EUR/MWh
                'import_limit': electricity_dict['import_limit'] * timestep, # MW
                'export_limit': electricity_dict['export_limit'] * timestep # MW
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def hydrogen_carrier(hydrogen_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'hydrogen.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'hydrogen']/8760*1000 if 'hydrogen' in nodes.columns else hydrogen_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': hydrogen_dict['import_price'], # EUR/t
                'demand': demand * timestep,     # t/h
                'export_price': hydrogen_dict['export_price'], # EUR/t
                'import_limit': hydrogen_dict['import_limit'] * timestep, # t/h
                'export_limit': hydrogen_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def hydrogen_transported_carrier(hydrogen_transported_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'hydrogen_transported.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'hydrogen_transported']/8760*1000 if 'hydrogen_transported' in nodes.columns else hydrogen_transported_dict['demand'] # convert from kt/y to t/h
            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': hydrogen_transported_dict['import_price'], # EUR/t
                'demand': demand * timestep,     # t/h
                'export_price': hydrogen_transported_dict['export_price'], # EUR/t
                'import_limit': hydrogen_transported_dict['import_limit'] * timestep, # t/h
                'export_limit': hydrogen_transported_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def ammonia_carrier(ammonia_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'ammonia.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'ammonia']/8760*1000 if 'ammonia' in nodes.columns else ammonia_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': ammonia_dict['import_price'], # EUR/t
                'demand': demand * timestep,     # t/h
                'export_price': ammonia_dict['export_price'], # EUR/t
                'import_limit': ammonia_dict['import_limit'] * timestep, # t/h
                'export_limit': ammonia_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def ethylene_carrier(ethylene_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'ethylene.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'ethylene']/8760*1000 if 'ethylene' in nodes.columns else ethylene_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': ethylene_dict['import_price'], # EUR/t
                'demand': demand * timestep,     # t/h
                'export_price': ethylene_dict['export_price'], # EUR/t
                'import_limit': ethylene_dict['import_limit'] * timestep, # t/h
                'export_limit': ethylene_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def propylene_carrier(propylene_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'propylene.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'propylene']/8760*1000 if 'propylene' in nodes.columns else propylene_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': propylene_dict['import_price'], # EUR/t
                'demand': demand * timestep,     # t/h
                'export_price': propylene_dict['export_price'], # EUR/t
                'import_limit': propylene_dict['import_limit'] * timestep, # t/h
                'export_limit': propylene_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def methanol_carrier(methanol_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'methanol.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'methanol']/8760*1000 if 'methanol' in nodes.columns else methanol_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': methanol_dict['import_price'], # EUR/t
                'demand': demand * timestep,     # t/h
                'export_price': methanol_dict['export_price'], # EUR/t
                'import_limit': methanol_dict['import_limit'] * timestep, # t/h
                'export_limit': methanol_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def CO2_carrier(CO2_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'CO2.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'CO2']/8760*1000 if 'CO2' in nodes.columns else CO2_dict['demand'] # convert from kt/y to t/h
            
            yearly_export_limit = CO2_dict['export_limit'][plant] if plant in CO2_dict['export_limit'] else 0 # t/h, if the plant is not in the dict, we consider that it has no export potential
            export_limit = yearly_export_limit/8760*1e6 # convert from Mt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': CO2_dict['import_price'], # EUR/t
                'demand': demand * timestep,     # t/h
                'export_price': CO2_dict['export_price'], # EUR/t
                'import_limit': CO2_dict['import_limit'] * timestep, # t/h
                'export_limit': export_limit * timestep, # t/h
                'export_emission_factor': CO2_dict['export_emission_factor'] # tCO2/t feedstock
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def methane_carrier(methane_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    country_dict = {
        'Belgium': 'BE',
        'Germany': 'DE',
        'Netherlands': 'NL'}

    LHV = 13.9 # MWh/t

    prices_data = pd.read_excel(project_folder / 'data' / 'carriers_data' / 'methane' / 'methane_prices.xlsx', sheet_name='prices', index_col=0)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'methane.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'methane']/8760*1000 if 'methane' in nodes.columns else methane_dict['demand'] # convert from kt/y to t/h
            country = nodes.loc[plant, 'country']
            country_code = country_dict[country]
            price = float(prices_data.loc[country_code, int(methane_dict['year'])]) * LHV # convert from EUR/MWh to EUR/t

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': price, # EUR/t
                'demand': demand * timestep,     # t/h
                'export_price': methane_dict['export_price'], # EUR/t
                'import_limit': methane_dict['import_limit'] * timestep, # t/h
                'export_limit': methane_dict['export_limit'] * timestep, # t/h
                'import_emission_factor': methane_dict['import_emission_factor'] # tCO2/t feedstock
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def steam_carrier(steam_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'steam.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'steam']/8760 if 'steam' in nodes.columns else steam_dict['demand'] # convert from GWh/y to MW

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': steam_dict['import_price'], # EUR/MWh
                'demand': demand * timestep, # MW
                'export_price': steam_dict['export_price'], # EUR/MWh
                'import_limit': steam_dict['import_limit'] * timestep, # MW
                'export_limit': steam_dict['export_limit'] * timestep # MW
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def biomass_carrier(biomass_dict, resolution):

    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    country_dict = {
    'Belgium': 'BE',
    'Germany': 'DE',
    'Netherlands': 'NL'}

    LHV = 15.5 # GJ/t, consider an average LHV for biomass feedstocks

    prices_data = pd.read_excel(project_folder / 'data' / 'carriers_data' / 'biomass' / 'biomass_prices.xlsx', sheet_name=biomass_dict['scenario'], index_col=0)
    availability_data = pd.read_excel(project_folder / 'data' / 'carriers_data' / 'biomass' / 'biomass_availability.xlsx', sheet_name=biomass_dict['scenario'], index_col=0)

    with pd.ExcelWriter(Path(carriers_path) / 'biomass.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            country = nodes.loc[plant, 'country']
            country_code = country_dict[country]

            # Prices are in a dataframe with country codes as index and years as columns, we take the column corresponding to the year of interest and the row corresponding to the country of the plant
            price = float(prices_data.loc[country_code, int(biomass_dict['year'])])
            price = price * LHV # convert from EUR/GJ to EUR/t

            yearly_availability = biomass_dict['import_limit_share'] /100 * float(availability_data.loc[country_code, int(biomass_dict['year'])]) # in kt/y
            if country_code == 'BE': # 2 nodes
                import_limit = yearly_availability * 1e6 / LHV / 8760 / 2 # convert from PJ/y to t/h
            elif country_code == 'NL': # 4 nodes
                import_limit = yearly_availability * 1e6 / LHV / 8760 / 4 # convert from PJ/y to t/h
            else: # 1 node
                import_limit = yearly_availability * 1e6 / LHV / 8760 # convert from PJ/y to t/h


            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': price, # EUR/t
                'demand': biomass_dict['demand'] * timestep, # t/h
                'export_price': biomass_dict['export_price'], # EUR/t
                'import_limit': import_limit * timestep, # t/h
                'export_limit': biomass_dict['export_limit'] * timestep, # t/h
                'import_emission_factor': biomass_dict['import_emission_factor'] # tCO2/t feedstock
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def methane_bio_carrier(methane_bio_dict, resolution):

    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    country_dict = {
        'Belgium': 'BE',
        'Germany': 'DE',
        'Netherlands': 'NL'
    }

    conversion_factor = 0.7 # kg/(cubic_meter)

    availability_data = pd.read_excel(project_folder / 'data' / 'carriers_data' / 'methane_bio' / 'methane_bio_availability.xlsx', sheet_name='availability', index_col=0)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'methane_bio.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            country = nodes.loc[plant, 'country']
            country_code = country_dict[country]

            demand = nodes.loc[plant, 'methane_bio']/8760*1000 if 'methane_bio' in nodes.columns else methane_bio_dict['demand'] # convert from kt/y to t/h
            yearly_potential = methane_bio_dict['import_limit_share'] /100 * float(availability_data.loc[country_code, int(methane_bio_dict['year'])]) # in bcm/y
            import_limit = yearly_potential * conversion_factor * 1e9 / 1e3 / 8760 # convert from bcm/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': methane_bio_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': methane_bio_dict['export_price'], # EUR/t
                'import_limit': import_limit * timestep, # t/h
                'export_limit': methane_bio_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def feedgas_carrier(feedgas_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    country_dict = {
        'Belgium': 'BE',
        'Germany': 'DE',
        'Netherlands': 'NL'}

    LHV = 13.9 # MWh/t

    # only for export
    prices_data = pd.read_excel(project_folder / 'data' / 'carriers_data' / 'methane' / 'methane_prices.xlsx', sheet_name='prices', index_col=0)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'feedgas.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'feedgas']/8760*1000 if 'feedgas' in nodes.columns else feedgas_dict['demand'] # convert from kt/y to t/h
            country = nodes.loc[plant, 'country']
            country_code = country_dict[country]
            price = float(prices_data.loc[country_code, int(methane_dict['year'])]) * LHV # convert from EUR/MWh to EUR/t

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': feedgas_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': feedgas_dict['export_price_percentage'] / 100 * price, # EUR/t
                'import_limit': feedgas_dict['import_limit'] * timestep, # t/h
                'export_limit': feedgas_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def syngas_carrier(syngas_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'syngas.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'syngas']/8760*1000 if 'syngas' in nodes.columns else syngas_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': syngas_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': syngas_dict['export_price'], # EUR/t
                'import_limit': syngas_dict['import_limit'] * timestep, # t/h
                'export_limit': syngas_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def crackergas_carrier(crackergas_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'crackergas.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'crackergas']/8760*1000 if 'crackergas' in nodes.columns else crackergas_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': crackergas_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': crackergas_dict['export_price'], # EUR/t
                'import_limit': crackergas_dict['import_limit'] * timestep, # t/h
                'export_limit': crackergas_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def olefins_carrier(olefins_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'olefins.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'olefins']/8760*1000 if 'olefins' in nodes.columns else olefins_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': olefins_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': olefins_dict['export_price'], # EUR/t
                'import_limit': olefins_dict['import_limit'] * timestep, # t/h
                'export_limit': olefins_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def naphtha_carrier(naphtha_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'naphtha.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'naphtha']/8760*1000 if 'naphtha' in nodes.columns else naphtha_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': naphtha_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': naphtha_dict['export_price'], # EUR/t
                'import_limit': naphtha_dict['import_limit'] * timestep, # t/h
                'export_limit': naphtha_dict['export_limit'] * timestep, # t/h
                'import_emission_factor': naphtha_dict['import_emission_factor'] # tCO2/t feedstock
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def HBfeed_carrier(HB_feed_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'HBfeed.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'HBfeed']/8760*1000 if 'HBfeed' in nodes.columns else HB_feed_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': HB_feed_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': HB_feed_dict['export_price'], # EUR/t
                'import_limit': HB_feed_dict['import_limit'] * timestep, # t/h
                'export_limit': HB_feed_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def nitrogen_carrier(nitrogen_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'nitrogen.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'nitrogen']/8760*1000 if 'nitrogen' in nodes.columns else nitrogen_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': nitrogen_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': nitrogen_dict['export_price'], # EUR/t
                'import_limit': nitrogen_dict['import_limit'] * timestep, # t/h
                'export_limit': nitrogen_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def MPW_carrier(MPW_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    country_dict = {
    'Belgium': 'BE',
    'Germany': 'DE',
    'Netherlands': 'NL'}

    prices_data = pd.read_excel(project_folder / 'data' / 'carriers_data' / 'MPW' / 'MPW_prices.xlsx', sheet_name=MPW_dict['scenario'], index_col=0)
    availability_data = pd.read_excel(project_folder / 'data' / 'carriers_data' / 'MPW' / 'MPW_availability.xlsx', sheet_name=MPW_dict['scenario'], index_col=0)

    with pd.ExcelWriter(Path(carriers_path) / 'MPW.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            country = nodes.loc[plant, 'country']
            country_code = country_dict[country]

            # Prices are in a dataframe with country codes as index and years as columns, we take the column corresponding to the year of interest and the row corresponding to the country of the plant
            price = float(prices_data.loc[country_code, int(MPW_dict['year'])])

            yearly_availability = MPW_dict['import_limit_share'] /100 * float(availability_data.loc[country_code, int(MPW_dict['year'])]) # in kt/y
            if country_code == 'BE': # 2 nodes
                import_limit = yearly_availability * 1000 / 8760 / 2 # convert from kt/y to t/h
            elif country_code == 'NL': # 4 nodes
                import_limit = yearly_availability * 1000 / 8760 / 4 # convert from kt/y to t/h
            else: # 1 node  
                import_limit = yearly_availability * 1000 / 8760 # convert from kt/y to t/h
            
            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': price, # EUR/t
                'demand': MPW_dict['demand'] * timestep, # t/h
                'export_price': MPW_dict['export_price'], # EUR/t
                'import_limit': import_limit * timestep, # t/h
                'export_limit': MPW_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def naphtha_bio_carrier(naphtha_bio_dict, resolution):

    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    country_dict = {
        'Belgium': 'BE',
        'Germany': 'DE',
        'Netherlands': 'NL'}
    
    availability_data = pd.read_excel(project_folder / 'data' / 'carriers_data' / 'naphtha_bio' / 'naphtha_bio_availability.xlsx', sheet_name=naphtha_bio_dict['import_limit_scenario'], index_col=0)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'naphtha_bio.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            country = nodes.loc[plant, 'country']
            country_code = country_dict[country]

            demand = nodes.loc[plant, 'naphtha_bio']/8760*1000 if 'naphtha_bio' in nodes.columns else naphtha_bio_dict['demand'] # convert from kt/y to t/h

            yearly_availability = float(availability_data.loc[country_code, int(naphtha_bio_dict['year'])]) # in kt/y
            import_limit = yearly_availability * 1000 / 8760 # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': naphtha_bio_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': naphtha_bio_dict['export_price'], # EUR/t
                'import_limit': import_limit * timestep, # t/h
                'export_limit': naphtha_bio_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def CO2captured_carrier(CO2captured_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'CO2captured.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'CO2captured']/8760*1000 if 'CO2captured' in nodes.columns else CO2captured_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': CO2captured_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': CO2captured_dict['export_price'], # EUR/t
                'import_limit': CO2captured_dict['import_limit'] * timestep, # t/h
                'export_limit': CO2captured_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def cracking_byproducts_carrier(cracking_byproducts_dict, resolution):
    case_study_folder, project_folder = folders_setup()
    nodes = get_nodes(case_study_folder)
    carriers_path = carriers_folder_setup(case_study_folder)
    full_index, timestep = create_full_index(resolution)

    # Write excel file
    with pd.ExcelWriter(Path(carriers_path) / 'cracking_byproducts.xlsx', engine='openpyxl', mode='w') as writer:
        for plant in nodes.index:

            demand = nodes.loc[plant, 'cracking_byproducts']/8760*1000 if 'cracking_byproducts' in nodes.columns else cracking_byproducts_dict['demand'] # convert from kt/y to t/h

            # Create DataFrame for the plant
            df = pd.DataFrame({
                'import_price': cracking_byproducts_dict['import_price'], # EUR/t
                'demand': demand * timestep, # t/h
                'export_price': cracking_byproducts_dict['export_price'], # EUR/t
                'import_limit': cracking_byproducts_dict['import_limit'] * timestep, # t/h
                'export_limit': cracking_byproducts_dict['export_limit'] * timestep # t/h
            }, index=full_index)

            # Save to dedicated sheet
            df.to_excel(writer, sheet_name=plant[:31], index_label='datetime')

    return

def folders_setup():
    try:
        # Try to get the notebook path from the VS Code variable
        notebook_path = __vsc_ipynb_file__
        case_study_folder = os.path.dirname(notebook_path)
    except NameError:
        # If the variable is not defined, use the current working directory
        case_study_folder = os.getcwd()
        print('Warning: __vsc_ipynb_file__ not found. Using current working directory as base path.')

    # Find folder for a one-level-up directory
    project_folder = Path(case_study_folder).parent.parent.parent

    return case_study_folder, project_folder

def get_nodes(case_study_folder):
    # Take nodes from the case study folder
    nodes = pd.read_csv(Path(case_study_folder) / 'plants_clusters_summary.csv', sep=',', index_col=0)    
    return nodes
   
def carriers_folder_setup(case_study_folder):
    carriers_path = Path(case_study_folder) / 'carriers'

    return carriers_path

def create_full_index(resolution):
    full_index = pd.date_range(start='2025-01-01 00:00:00', end='2025-12-31 23:00:00', freq=resolution)
    if resolution == 'h':
        timestep = 1
    elif resolution == '2h':
        timestep = 2
    elif resolution == '3h':
        timestep = 3
    elif resolution == '4h':
        timestep = 4
    else:
        raise ValueError(f"Unexpected resolution: {resolution}. Cannot determine timestep.")
    return full_index, timestep

def define_resolution(electricity_dict):
    case_study_folder, project_folder = folders_setup()
    carriers_path = carriers_folder_setup(case_study_folder)

    if electricity_dict['resolution'] == '1h':
        if electricity_dict['year'] == '2025':
            prices_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/prices_AC_allin_2025.csv', sep=',', index_col=0)
        else:
            prices_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/prices_AC_allin_{electricity_dict["scenario"]}_{electricity_dict["year"]}.csv', sep=',', index_col=0)
    elif electricity_dict['resolution'] == '2h':
        if electricity_dict['year'] == '2025':
            prices_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/prices_AC_allin_2025_2h.csv', sep=',', index_col=0)
        else:
            prices_data = pd.read_csv(project_folder / f'data/carriers_data/electricity/prices_AC_allin_{electricity_dict["scenario"]}_{electricity_dict["year"]}_2h.csv', sep=',', index_col=0)
    length = prices_data.shape[0]
    
    if length == 8760:
        resolution = 'h'
    elif length == 8760/2:
        resolution = '2h'
    elif length == 8760/3:
        resolution = '3h'
    elif length == 8760/4:
        resolution = '4h'
    else:
        raise ValueError(f"Unexpected number of rows in electricity prices data: {length}. Cannot determine resolution.")
    
    return resolution

    

resolution = define_resolution(electricity_grid_dict)
carbon_costs_definition(carbon_costs_dict, resolution)
electricity_grid_carrier(electricity_grid_dict, resolution)
electricity_carrier(electricity_dict, resolution)
hydrogen_carrier(hydrogen_dict, resolution)
hydrogen_transported_carrier(hydrogen_transported_dict, resolution)
ammonia_carrier(ammonia_dict, resolution)
ethylene_carrier(ethylene_dict, resolution)
propylene_carrier(propylene_dict, resolution)
methanol_carrier(methanol_dict, resolution)
CO2_carrier(CO2_dict, resolution)
methane_carrier(methane_dict, resolution)
# heat_carrier(heat_dict)
steam_carrier(steam_dict, resolution)
biomass_carrier(biomass_dict, resolution)
methane_bio_carrier(methane_bio_dict, resolution)
feedgas_carrier(feedgas_dict, resolution)
syngas_carrier(syngas_dict, resolution)
crackergas_carrier(crackergas_dict, resolution)
olefins_carrier(olefins_dict, resolution)
naphtha_carrier(naphtha_dict, resolution)
HBfeed_carrier(HBfeed_dict, resolution)
nitrogen_carrier(nitrogen_dict, resolution)
MPW_carrier(MPW_dict, resolution)
naphtha_bio_carrier(naphtha_bio_dict, resolution)
CO2captured_carrier(CO2captured_dict, resolution)
# CO2_DAC_carrier(CO2_DAC_dict)
cracking_byproducts_carrier(cracking_byproducts_dict, resolution)
```

```python
import h5py
import numpy as np
import pandas as pd
from openpyxl import Workbook, load_workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter
from pathlib import Path
import os

##### Files names
from_case_study   = '2040'
xlxs_ref  = r"data\technologies_fac_simile.xlsx"   # per prendere l'ordine delle colonne
output_name  = r"technologies.xlsx"
year      = "2022"

##### Read sizes from h5 and create formatted excel output
def _to_scalar(x):
    x = np.asarray(x)
    return float(x.reshape(-1)[0]) if x.size == 1 else float("nan")

##### Folders setup
def folders_setup():
    '''
    This function sets up the paths for the case study folder and the project folder.
    It first tries to get the path of the current notebook using the VS Code variable __vsc_ipynb_file__. If this variable is not defined (e.g., if the code is run in a different environment), it falls back to using the current working directory.
    Then, it defines the project folder as the parent of the parent of the case study folder (assuming the case study folder is located at project_folder/case_study_folder).
    Returns:
        case_study_folder (str): The path to the case study folder.
        project_folder (str): The path to the project folder.
    '''
    try:
        # Try to get the notebook path from the VS Code variable
        notebook_path = __vsc_ipynb_file__
        case_study_folder = os.path.dirname(notebook_path)
    except NameError:
        # If the variable is not defined, use the current working directory
        case_study_folder = os.getcwd()
        print('Warning: __vsc_ipynb_file__ not found. Using current working directory as base path.')

    project_folder = Path(case_study_folder).parent.parent

    return case_study_folder, project_folder

def electricity_connection_correction(ws_existing, ws_new):
    """
    Manually correct electricity grid connection data.

    In the existing sheet:
    - all electricity grid connection columns are set to 0.

    In the new sheet:
    - electricity grid connection columns are set to 1 only for the
      corresponding node-country connection;
    - all other electricity grid connection columns are set to 0.
    """

    electricity_connections = [
        "electricity_grid_connection_BE",
        "electricity_grid_connection_DE",
        "electricity_grid_connection_NL",
    ]

    couples_node_electricity_connections = {
        "BEL1": ["electricity_grid_connection_BE"],
        "BEL2": ["electricity_grid_connection_BE"],
        "DE":   ["electricity_grid_connection_DE"],
        "NL1":  ["electricity_grid_connection_NL"],
        "NL2":  ["electricity_grid_connection_NL"],
        "NL3":  ["electricity_grid_connection_NL"],
        "NL4":  ["electricity_grid_connection_NL"],
    }

    # Identify electricity connection columns from header
    existing_header = {
        cell.value: cell.column
        for cell in ws_existing[1]
        if cell.value is not None
    }

    new_header = {
        cell.value: cell.column
        for cell in ws_new[1]
        if cell.value is not None
    }

    existing_elec_cols = [
        existing_header[tech]
        for tech in electricity_connections
        if tech in existing_header
    ]

    new_elec_cols = [
        new_header[tech]
        for tech in electricity_connections
        if tech in new_header
    ]

    # Existing sheet: all electricity connection values to 0
    for row_idx in range(2, ws_existing.max_row + 1):
        for col_idx in existing_elec_cols:
            ws_existing.cell(row=row_idx, column=col_idx).value = 0

    # New sheet: only node-matching electricity connection values to 1
    for row_idx in range(2, ws_new.max_row + 1):
        node = ws_new.cell(row=row_idx, column=1).value
        allowed_connections = couples_node_electricity_connections.get(node, [])

        for col_idx in new_elec_cols:
            tech = ws_new.cell(row=1, column=col_idx).value
            ws_new.cell(row=row_idx, column=col_idx).value = (
                1 if tech in allowed_connections else 0
            )

    return ws_existing, ws_new

def H2_network_connection_correction(ws_existing):
    """
    Just to manually correct data on H2 connections
    """
    
    H2_connections = [
        "H2_network_connection_in",
        "H2_network_connection_out"
    ]

    # In existing sheet, set all values in columns of H2 connections to 0
    for row in ws_existing.iter_rows(min_row=2, min_col=2, max_col=ws_existing.max_column):
        for cell in row:
            if cell.column_letter in [get_column_letter(i) for i, tech in enumerate(ws_existing[1], start=1) if tech.value in H2_connections]:
                cell.value = 0

    return ws_existing

case_study_folder, project_folder = folders_setup()

results_path = project_folder / 'output' / from_case_study / 'optimization_results.h5'

base = f"design/nodes/{year}"
rows = []
with h5py.File(results_path, "r") as f:
    for node in f[base].keys():
        for comp in f[f"{base}/{node}"].keys():
            ds_path = f"{base}/{node}/{comp}/size"
            size = _to_scalar(f[ds_path][()]) if ds_path in f else 0.0
            component_name = comp[:-9] if comp.endswith("_existing") else comp
            rows.append({"node": node, "component": component_name, "size": size})

sizes_df = pd.DataFrame(rows)
sizes_df = sizes_df.groupby(["node", "component"], as_index=False)["size"].sum()
pivot = sizes_df.pivot(index="node", columns="component", values="size").fillna(0)

##### Read reference technologies and create output table
wb_ref    = load_workbook(xlxs_ref, read_only=True)
ws_ref    = wb_ref["existing"]
header_ref = list(next(ws_ref.iter_rows(min_row=1, max_row=1, values_only=True)))
nodes_ref  = [row[0] for row in ws_ref.iter_rows(min_row=2, values_only=True)]
tech_cols  = header_ref[1:]

##### Create output DataFrame with all nodes and technologies, filling missing values with 0
output_rows = []
for node in nodes_ref:
    row = {"cluster": node}
    for tech in tech_cols:
        if tech in pivot.columns and node in pivot.index:
            row[tech] = round(pivot.loc[node, tech], 2)
        else:
            row[tech] = 0
    output_rows.append(row)

out_df = pd.DataFrame(output_rows)

##### Write formatted excel output
def write_formatted_sheet(wb, sheet_name, df, all_cols):
    if sheet_name in wb.sheetnames:
        ws = wb[sheet_name]
    else:
        ws = wb.create_sheet(sheet_name)

    header_font  = Font(name="Arial", bold=True, color="FFFFFF", size=10)
    header_fill  = PatternFill("solid", start_color="2E4057")
    data_font    = Font(name="Arial", size=10)
    alt_fill     = PatternFill("solid", start_color="EAF0FB")
    white_fill   = PatternFill("solid", start_color="FFFFFF")
    center_align = Alignment(horizontal="center", vertical="center")
    left_align   = Alignment(horizontal="left", vertical="center")
    thin   = Side(style="thin", color="CCCCCC")
    border = Border(left=thin, right=thin, top=thin, bottom=thin)

    for c_idx, col_name in enumerate(all_cols, start=1):
        cell = ws.cell(row=1, column=c_idx, value=col_name)
        cell.font = header_font
        cell.fill = header_fill
        cell.alignment = center_align
        cell.border = border

    for r_idx, row in enumerate(df.itertuples(index=False), start=2):
        alt = (r_idx % 2 == 0)
        for c_idx, val in enumerate(row, start=1):
            cell = ws.cell(row=r_idx, column=c_idx, value=val)
            cell.font = data_font
            cell.fill = alt_fill if alt else white_fill
            cell.alignment = left_align if c_idx == 1 else center_align
            cell.border = border
            if c_idx > 1:
                cell.number_format = '#,##0.00;-#,##0.00;0.00'

    ws.column_dimensions["A"].width = 12
    for c_idx in range(2, len(all_cols) + 1):
        ws.column_dimensions[get_column_letter(c_idx)].width = max(len(all_cols[c_idx-1]) + 2, 10)

    ws.row_dimensions[1].height = 18
    ws.freeze_panes = "B2"

    return ws


wb = Workbook()

ws_existing = wb.active
ws_existing.title = "existing"

all_cols = ["cluster"] + tech_cols

write_formatted_sheet(wb, "existing", out_df, all_cols)

new_df = pd.DataFrame(
    [{"cluster": node, **{tech: 1 for tech in tech_cols}} for node in nodes_ref]
)

write_formatted_sheet(wb, "new", new_df, all_cols)

ws_existing = wb["existing"]
ws_new = wb["new"]

ws_existing, ws_new = electricity_connection_correction(ws_existing, ws_new)
ws_existing = H2_network_connection_correction(ws_existing)

wb.save(Path(case_study_folder) / output_name)
print("Saved:", output_name)
```
